# Pipeline Optimization

This notebook performs Delta Lake optimization and storage cleanup operations
to improve query performance and storage efficiency.

Optimization techniques used:
- OPTIMIZE
- VACUUM
- CACHE
- Partitioning concepts


```python
%sql

OPTIMIZE retail_silver.customers;
```

```python
%sql

OPTIMIZE retail_silver.orders;
```

```python
%sql

OPTIMIZE retail_silver.order_items;
```

# OPTIMIZE

OPTIMIZE compacts small files into larger optimized files,
improving query performance in Delta Lake tables.

```python
%sql

VACUUM retail_silver.customers RETAIN 168 HOURS;
```

```python
%sql

VACUUM retail_silver.orders RETAIN 168 HOURS;
```

# VACUUM

VACUUM removes old unused files from Delta tables
to reduce storage usage and improve maintenance.

```python
customers = spark.table("retail_silver.customers")

customers.show(5)
```
