```python
# %sql
# CREATE DATABASE IF NOT EXISTS retail_silver
spark.sql("CREATE DATABASE IF NOT EXISTS retail_silver")
```

```python
customers.write.format('delta').mode('overwrite').saveAsTable('retail_silver.customers')

orders.write.format('delta').mode('overwrite').saveAsTable('retail_silver.orders')

categories.write.format('delta').mode('overwrite').saveAsTable('retail_silver.categories')

departments.write.format('delta').mode('overwrite').saveAsTable('retail_silver.departments')

products.write.format('delta').mode('overwrite').saveAsTable('retail_silver.products')

order_items.write.format('delta').mode('overwrite').saveAsTable('retail_silver.order_items')
```
